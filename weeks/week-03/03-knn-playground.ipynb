# Week 3: Let the neighbours vote

**Question:** Can nearby examples decide the label of a new point?

We build intuition for k-nearest neighbours before treating it as an API call.

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import make_moons
from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score, ConfusionMatrixDisplay
```

```python
X, y = make_moons(n_samples=300, noise=0.25, random_state=7)
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=.30, random_state=7, stratify=y)
plt.figure(figsize=(7, 5))
plt.scatter(X_train[:, 0], X_train[:, 1], c=y_train, cmap='coolwarm', edgecolor='white')
plt.title('Training examples'); plt.show()
```

### Predict

What will happen when `k=1`? What will happen if `k` is almost as large as the training set?

```python
def show_knn(k):
    model = KNeighborsClassifier(n_neighbors=k).fit(X_train, y_train)
    x1 = np.linspace(X[:, 0].min()-.5, X[:, 0].max()+.5, 250)
    x2 = np.linspace(X[:, 1].min()-.5, X[:, 1].max()+.5, 250)
    xx1, xx2 = np.meshgrid(x1, x2)
    grid = np.column_stack((xx1.ravel(), xx2.ravel()))
    regions = model.predict(grid).reshape(xx1.shape)
    plt.figure(figsize=(7, 5))
    plt.contourf(xx1, xx2, regions, alpha=.18, cmap='coolwarm')
    plt.scatter(X_train[:, 0], X_train[:, 1], c=y_train, cmap='coolwarm', edgecolor='white', s=35)
    train_acc = accuracy_score(y_train, model.predict(X_train))
    test_acc = accuracy_score(y_test, model.predict(X_test))
    plt.title(f'k={k} | train={train_acc:.1%} | test={test_acc:.1%}')
    plt.show()
    return model

model = show_knn(5)
```

## Experiment

Try `k = 1, 3, 5, 15, 50, 150`. Describe how the boundary and the gap between training and test accuracy change.

```python
model = show_knn(1)  # change k
```

## Evaluation is more than one number

Accuracy says how often the prediction was correct. A confusion matrix shows which classes were confused.

```python
ConfusionMatrixDisplay.from_estimator(model, X_test, y_test, cmap='Blues')
plt.show()
```

## Deliberate failure: feature scale

kNN depends directly on distance. If one feature is measured in thousands and another between zero and one, the large-scale feature may dominate even when it is not more important.
