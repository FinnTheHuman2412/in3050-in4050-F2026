# Week 1: From numbers to a decision boundary

A machine-learning model does not see a person, email, image, or robot. It sees **numbers**. Today we build the path from one list of numbers to a visible classification rule.

## Learning goals

## 1. One object becomes one vector

Imagine classifying a fictional creature using two features: speed and strength. A creature with speed 2 and strength 4 becomes $x=[2,4]$. The order matters because each position has a meaning.

```python
import numpy as np
import matplotlib.pyplot as plt

x = np.array([2.0, 4.0])
x
```

### 🧠 Predict
If we record three creatures, each with speed and strength, should the dataset matrix have shape `(3, 2)` or `(2, 3)`? Explain what every row and column means before running the next cell.

```python
X = np.array([[2.0, 4.0],
              [5.0, 1.0],
              [4.0, 3.0]])
print('shape:', X.shape)
print(X)
```

## 2. A dot product becomes a score

Let $w=[1,-1]$ and $b=0$. Then $w^Tx+b$ rewards speed and penalizes strength. Positive and negative scores can be mapped to two classes.

```python
w = np.array([1.0, -1.0])
b = 0.0
scores = X @ w + b
predictions = (scores >= 0).astype(int)
for example, score, prediction in zip(X, scores, predictions):
    print(f'{example} -> score={score: .1f}, class={prediction}')
```

## 3. Make the rule visible

All points satisfying $w_1x_1+w_2x_2+b=0$ lie on the boundary. On one side the score is positive; on the other it is negative.

```python
def plot_boundary(w, b, X):
    x1 = np.linspace(0, 6, 200)
    x2 = -(w[0] * x1 + b) / w[1]
    scores = X @ w + b
    classes = (scores >= 0).astype(int)
    plt.figure(figsize=(7, 5))
    plt.scatter(X[:, 0], X[:, 1], c=classes, cmap='coolwarm', s=100, edgecolor='black')
    plt.plot(x1, x2, color='black', label=r'$w^Tx+b=0$')
    plt.xlim(0, 6); plt.ylim(0, 6)
    plt.xlabel('speed ($x_1$)'); plt.ylabel('strength ($x_2$)')
    plt.grid(alpha=.25); plt.legend(); plt.show()

plot_boundary(w, b, X)
```

## 4. Classification or regression?

The score itself is continuous. Turning it into class 0 or 1 makes this a classifier. If the score were used directly to predict a temperature or price, it would be regression. The same linear algebra can support different tasks.
