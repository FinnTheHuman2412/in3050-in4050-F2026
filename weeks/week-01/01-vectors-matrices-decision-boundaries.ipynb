# Week 1: From numbers to a decision boundary

A model does not see a creature, email or image. It sees numbers. Today we follow the path from one object to one visible classification rule.

## Learning goals

Interpret vectors and matrices, compute a dot product, and explain why $w^Tx+b=0$ is a linear decision boundary.

```python
import numpy as np
import matplotlib.pyplot as plt
```

## 1. One object becomes one vector

Imagine classifying a fictional creature using **speed** and **strength**. A creature with speed 2 and strength 4 becomes $x=[2,4]$. The order matters because each position has a meaning.

```python
x = np.array([2.0, 4.0])
print('one creature:', x)
print('shape:', x.shape)
```

### Predict

If we record 100 creatures with these two features, should the dataset shape be `(100, 2)` or `(2, 100)`? What does every row and column mean?

```python
X = np.array([[2.0, 4.0],
              [5.0, 1.0],
              [4.0, 3.0],
              [1.0, 5.0],
              [5.0, 4.0]])
print('dataset shape:', X.shape)
print(X)
```

## 2. A dot product becomes a score

Let $w=[1,-1]$ and $b=0$. The score rewards speed and penalizes strength. A non-negative score becomes class 1.

```python
w = np.array([1.0, -1.0])
b = 0.0
scores = X @ w + b
predictions = (scores >= 0).astype(int)
for example, score, prediction in zip(X, scores, predictions):
    print(f'{example} -> score={score: .1f}, class={prediction}')
```

## 3. Make the rule visible

All points satisfying $w_1x_1+w_2x_2+b=0$ lie on the boundary. The two sides receive different predictions.

```python
def plot_boundary(w, b, X):
    x1 = np.linspace(0, 6, 250)
    if np.isclose(w[1], 0):
        boundary = -b / w[0]
        plt.axvline(boundary, color='black', label=r'$w^Tx+b=0$')
    else:
        x2 = -(w[0] * x1 + b) / w[1]
        plt.plot(x1, x2, color='black', label=r'$w^Tx+b=0$')
    scores = X @ w + b
    classes = (scores >= 0).astype(int)
    plt.scatter(X[:, 0], X[:, 1], c=classes, cmap='coolwarm', s=110, edgecolor='black')
    for index, point in enumerate(X):
        plt.annotate(str(index), point + 0.08)
    plt.xlim(0, 6); plt.ylim(0, 6)
    plt.xlabel('speed ($x_1$)'); plt.ylabel('strength ($x_2$)')
    plt.grid(alpha=.25); plt.legend(); plt.show()

plt.figure(figsize=(7, 5))
plot_boundary(w, b, X)
```

## 4. Pair experiment

Change **one thing at a time** and predict before running.

## 5. Deliberately break the model: XOR

Can one straight line separate the red points from the blue points below?

```python
X_xor = np.array([[0, 0], [0, 1], [1, 0], [1, 1]])
y_xor = np.array([0, 1, 1, 0])
plt.figure(figsize=(5, 5))
plt.scatter(X_xor[:, 0], X_xor[:, 1], c=y_xor, cmap='coolwarm', s=180, edgecolor='black')
plt.xticks([0, 1]); plt.yticks([0, 1]); plt.grid(alpha=.25)
plt.title('XOR: not linearly separable'); plt.show()
```

Changing the weights can move or rotate a line, but it cannot bend the line. XOR therefore cannot be solved by a single linear decision boundary. This limitation motivates feature transformations and multi-layer neural networks.
