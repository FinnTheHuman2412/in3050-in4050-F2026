# Week 2: Search and optimization

**Question:** How do we find a good solution when checking everything is expensive?

We compare exhaustive search and gradient ascent on the same landscape.

```python
import numpy as np
import matplotlib.pyplot as plt
```

## 1. See the search landscape

The horizontal axis is the candidate solution. The vertical axis is its objective or fitness.

```python
def objective(x):
    return -x**4 + 2*x**3 + 2*x**2 - x

def gradient(x):
    return -4*x**3 + 6*x**2 + 4*x - 1

xx = np.linspace(-2, 3, 600)
plt.figure(figsize=(8, 4.5))
plt.plot(xx, objective(xx))
plt.axhline(0, color='black', linewidth=.7)
plt.xlabel('candidate x'); plt.ylabel('objective f(x)'); plt.grid(alpha=.25); plt.show()
```

### Predict

If gradient ascent starts at two different positions, must it finish at the same maximum?

```python
def gradient_ascent(start, step_size=0.05, steps=30):
    path = [float(start)]
    for _ in range(steps):
        path.append(path[-1] + step_size * gradient(path[-1]))
    return np.array(path)

for start in [-1.5, 0.0, 2.2]:
    path = gradient_ascent(start)
    plt.plot(xx, objective(xx), color='lightgray')
    plt.plot(path, objective(path), 'o-', label=f'start={start}')
    plt.legend(); plt.grid(alpha=.2); plt.show()
```

## 2. Exhaustive search

Exhaustive search evaluates every allowed candidate. It is simple and reliable when the search space is small.

```python
candidates = np.arange(-2, 3.01, 0.5)
values = objective(candidates)
best_index = np.argmax(values)
print('best candidate:', candidates[best_index])
print('best value:', values[best_index])
plt.figure(figsize=(8, 4.5))
plt.plot(xx, objective(xx), color='gray')
plt.scatter(candidates, values, label='evaluated candidates')
plt.scatter(candidates[best_index], values[best_index], s=150, marker='*', label='best found')
plt.legend(); plt.grid(alpha=.25); plt.show()
```

## Deliberate failure

A gradient gives a locally useful direction; it does not promise the global optimum. A large step can also overshoot or diverge. An exhaustive method avoids local traps only if enumerating the search space is feasible.
