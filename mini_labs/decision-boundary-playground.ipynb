# Decision-boundary playground

**Question:** When is one straight line enough to classify the data?

Change the dataset while keeping the classifier fixed. The failure is the lesson.

```python
import matplotlib.pyplot as plt
import numpy as np
from sklearn.datasets import make_blobs, make_circles, make_moons
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score
```

## 1. Choose a world

Start with `blobs`. Before trying another dataset, predict whether a straight boundary will work.

```python
DATASET = 'blobs'  # Try: 'blobs', 'moons', 'circles'
NOISE = 0.18

if DATASET == 'blobs':
    X, y = make_blobs(n_samples=300, centers=2, cluster_std=1.15, random_state=7)
elif DATASET == 'moons':
    X, y = make_moons(n_samples=300, noise=NOISE, random_state=7)
elif DATASET == 'circles':
    X, y = make_circles(n_samples=300, noise=NOISE / 2, factor=0.45, random_state=7)
else:
    raise ValueError('Unknown dataset')

plt.figure(figsize=(7, 5))
plt.scatter(X[:, 0], X[:, 1], c=y, cmap='coolwarm', edgecolor='white', s=45)
plt.xlabel('$x_1$'); plt.ylabel('$x_2$'); plt.title(DATASET.title()); plt.show()
```

## 2. Fit one linear classifier

The model is deliberately unchanged. Only the geometry of the data changes.

```python
model = LogisticRegression().fit(X, y)
x1 = np.linspace(X[:, 0].min() - 0.5, X[:, 0].max() + 0.5, 250)
x2 = np.linspace(X[:, 1].min() - 0.5, X[:, 1].max() + 0.5, 250)
xx1, xx2 = np.meshgrid(x1, x2)
grid = np.column_stack((xx1.ravel(), xx2.ravel()))
prediction = model.predict(grid).reshape(xx1.shape)

plt.figure(figsize=(7, 5))
plt.contourf(xx1, xx2, prediction, alpha=0.18, cmap='coolwarm')
plt.contour(xx1, xx2, prediction, levels=[0.5], colors='black', linewidths=2)
plt.scatter(X[:, 0], X[:, 1], c=y, cmap='coolwarm', edgecolor='white', s=45)
plt.title(f'{DATASET.title()} | training accuracy = {accuracy_score(y, model.predict(X)):.1%}')
plt.xlabel('$x_1$'); plt.ylabel('$x_2$'); plt.show()
```

## Explain before moving on

1. What does the black line represent?
2. Why can changing the weights rotate and move the line, but never bend it?
3. Which dataset cannot be solved well by *any* single straight boundary?
4. Why does a high training accuracy not automatically prove good generalization?
